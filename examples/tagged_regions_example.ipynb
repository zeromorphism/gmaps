# Custom Tagged Regions Example

This notebook demonstrates the `TaggedRegion` layer for creating custom regions with metadata, tags, and links to Wikipedia-like sources.

```python
import gmaps
import gmaps.datasets

# Configure your API key (replace with your actual key)
# Get a key at https://console.developers.google.com/apis/credentials
gmaps.configure(api_key="YOUR_API_KEY")
```

## Basic Tagged Region

Let's create a simple region representing Golden Gate Park in San Francisco.

```python
# Create a map centered on San Francisco
m = gmaps.Map()

# Define Golden Gate Park as a rectangular region
golden_gate_park = gmaps.TaggedRegion(
    name="Golden Gate Park",
    coordinates=[
        (37.771, -122.511),
        (37.771, -122.453),
        (37.766, -122.453),
        (37.766, -122.511)
    ],
    description="Large urban park in San Francisco, spanning over 1,000 acres.",
    tags=["park", "recreation", "landmark", "urban-green-space"],
    wikipedia_url="https://en.wikipedia.org/wiki/Golden_Gate_Park",
    fill_color="green",
    fill_opacity=0.4,
    stroke_color="darkgreen",
    stroke_weight=2
)

m.add_layer(golden_gate_park)
m
```

## Multiple Regions with Custom Metadata

Create multiple regions with different properties and custom data.

```python
# Create a new map for multiple regions
m2 = gmaps.Map()

# Silicon Valley - Tech Hub
silicon_valley = gmaps.TaggedRegion(
    name="Silicon Valley",
    coordinates=[
        (37.4, -122.2),
        (37.4, -121.9),
        (37.3, -121.9),
        (37.3, -122.2)
    ],
    description="Global center for high technology and innovation.",
    tags=["technology", "innovation", "business"],
    data={
        "Founded": "1970s",
        "Famous Companies": "Google, Apple, Facebook, Tesla",
        "Industry": "Technology & Software"
    },
    wikipedia_url="https://en.wikipedia.org/wiki/Silicon_Valley",
    fill_color="#4285F4",
    fill_opacity=0.35
)

# San Francisco Bay
sf_bay = gmaps.TaggedRegion(
    name="San Francisco Bay",
    coordinates=[
        (37.85, -122.5),
        (37.85, -122.2),
        (37.45, -122.0),
        (37.45, -122.3)
    ],
    description="A shallow estuary in Northern California.",
    tags=["water", "estuary", "natural", "ecology"],
    data={
        "Type": "Estuary",
        "Area": "~1,600 km²",
        "Max Depth": "~110 ft"
    },
    wikipedia_url="https://en.wikipedia.org/wiki/San_Francisco_Bay",
    fill_color="#0066CC",
    fill_opacity=0.3,
    stroke_color="#003366"
)

# Napa Valley - Wine Region
napa_valley = gmaps.TaggedRegion(
    name="Napa Valley",
    coordinates=[
        (38.5, -122.5),
        (38.5, -122.3),
        (38.3, -122.3),
        (38.3, -122.5)
    ],
    description="World-renowned wine region north of San Francisco.",
    tags=["wine", "agriculture", "tourism", "gourmet"],
    data={
        "Primary Industry": "Wine Production",
        "Wineries": "~400+",
        "Established": "1960s (modern era)"
    },
    wikipedia_url="https://en.wikipedia.org/wiki/Napa_Valley_AVA",
    fill_color="#800020",
    fill_opacity=0.4,
    stroke_color="#4A0012"
)

# Add all regions to the map
m2.add_layer(silicon_valley)
m2.add_layer(sf_bay)
m2.add_layer(napa_valley)

m2
```

## Complex Polygon Region

Create a more complex polygon shape (not just a rectangle).

```python
# Create a map for complex shapes
m3 = gmaps.Map()

# San Francisco city boundaries (simplified)
sf_city = gmaps.TaggedRegion(
    name="San Francisco",
    coordinates=[
        (37.8, -122.52),
        (37.81, -122.48),
        (37.78, -122.38),
        (37.72, -122.38),
        (37.71, -122.42),
        (37.72, -122.50)
    ],
    description="The City by the Bay - cultural, commercial, and financial center.",
    tags=["city", "metropolitan", "culture", "tech-hub"],
    data={
        "Population": "~874,000",
        "Area": "~121 km²",
        "Founded": "1776",
        "County": "San Francisco County"
    },
    wikipedia_url="https://en.wikipedia.org/wiki/San_Francisco",
    fill_color="#FF4500",
    fill_opacity=0.3,
    stroke_color="#CC3700",
    stroke_weight=3
)

m3.add_layer(sf_city)
m3
```

## Customization Options

The `TaggedRegion` layer supports the following parameters:

**Required:**
- `name`: Name of the region (string)
- `coordinates`: List of (latitude, longitude) tuples (minimum 3 points)

**Optional:**
- `description`: Description text (string)
- `tags`: List of tag strings
- `data`: Dictionary of custom metadata
- `wikipedia_url`: URL to external source (string)
- `fill_color`: Polygon fill color (CSS color or hex)
- `fill_opacity`: Fill transparency (0.0 to 1.0)
- `stroke_color`: Border color (CSS color or hex)
- `stroke_opacity`: Border transparency (0.0 to 1.0)
- `stroke_weight`: Border width in pixels (integer)

## Use Cases

This feature is useful for:

- **Education**: Create interactive geography lessons

- **Environmental studies**: Mark conservation areas, habitats, or study regions
